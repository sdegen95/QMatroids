```julia
hyps = [matrix(GF(2),[1 0 0;0 0 1]),matrix(GF(2),[0 1 0;0 0 1]),matrix(GF(2),[1 1 0; 0 0 1])]
```

```julia
actfn = function(spaces,G)
        return Set([rref(C*transpose(matrix(G)))[2] for C in spaces])
    end
GLn = Oscar.general_linear_group(3, GF(2))

S_prime, emb_prime = oscar.stabilizer(GLn, Set(hyps), actfn);
```

```julia
Oscar.order(S_prime)
```

```julia
mat_S = elements(S_prime)[24]
mat_aut = elements(aut)[24]
#Set([rref(B*transpose(mat_S))[2] for B in hyps]) == Set(hyps)
E = actfn(Set(hyps),mat_S)
```

```julia
E==Set(hyps)
```

```julia
intersect(E,Set(hyps))
```

```julia
Stab = [g for g in GLn if actfn(Set(hyps),g)==Set(hyps)];
```

```julia
actfn(Set(hyps),Stab[2])==Set(hyps)
```

```julia
Set(Stab)==Set(elements(aut))
```
