# ***Enumeration of $q$-matroids on $\mathbb{F}_2^4$***
***

````julia
# Collect all different q-matroids isomorphism classes representatives of rank 2
Ext_F,x = finite_field(2,3,"x")
dim3_rank2_qm = [] 
matrix_list = [matrix(Ext_F,[1 0 0;0 1 0]),matrix(Ext_F,[1 0 0;0 1 x]),matrix(Ext_F,[1 0 x;0 1 x^2])]
for mat in matrix_list
    QM = q_matroid_from_matrix(mat)
    fs = Q_Matroid_Flats(QM)
    push!(dim3_rank2_qm,[QM,fs])
end
#dim3_rank2_qm
````

````julia
# Collect all lqs-selectors for all possible choices of q-matroids compute their corresponding extension q-matroids bases
all_ext_qms = []
for (id,elm) in enumerate(dim3_rank2_qm)
    println("$id. q-matroid of $(length(dim3_rank2_qm)) in progress")
    println("---------------------------------------------------------")
    lqs_selectors, hyps, ones_dict = Iter_DFS_enumV2(elm[1],true)
    #count = 0
    for i in range(1,length(lqs_selectors))
        Ext_QM = LqS_selector_to_q_matroid(elm[1],elm[2],hyps,lqs_selectors[i],ones_dict)
        push!(all_ext_qms,Ext_QM.bases)
        # ba = Ext_QM.bases
        # if Are_q_matroid_bases(ba)
        #     count += 1
        #     push!(all_ext_qms,ba)
        # end
    end
    # println("#lqs_selectors, #pushed bases: $(length(lqs_selectors))", ", ", "$count")
    # println("-------------------------------------")
end
````

````julia
length(all_ext_qms)
````

````julia
# Compute the isomorphism class from the list of all extension q-matroids bases
Isom_classes_from_basesV2(all_ext_qms)
````
